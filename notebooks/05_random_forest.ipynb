# 05. Random Forest Ensemble Classifier

Ensemble bagging classifier combining multiple decision trees to reduce variance and predict romantic compatibility.

```python
import sys
import os
import pandas as pd
sys.path.append(os.path.abspath('..'))

from src.data.preprocessing import CupidDataPreprocessor
from src.models.random_forest import CupidRandomForest
from src.evaluation.metrics import evaluate_classification_model, print_evaluation_summary

print("Modules importés avec succès.")
```

```python
print("Chargement et ingénierie des caractéristiques...")
preprocessor = CupidDataPreprocessor(test_size=0.2, random_state=42)
raw_path = '../data/raw/cupid_algorithm_dataset.csv' if os.path.exists('../data/raw/cupid_algorithm_dataset.csv') else 'data/raw/cupid_algorithm_dataset.csv'
X_train, X_test, y_train, y_test = preprocessor.prepare_train_test(raw_path)
print(f"X_train shape: {X_train.shape}, X_test shape: {X_test.shape}")
```

```python
print("Entraînement de la Random Forest (100 arbres)...")
model = CupidRandomForest(n_estimators=100, max_depth=14, n_jobs=-1, random_state=42)
model.fit(X_train, y_train)
print("Entraînement terminé avec succès !")
```

```python
metrics = evaluate_classification_model(model, X_test, y_test, 'Random Forest')
print_evaluation_summary(metrics)
```
