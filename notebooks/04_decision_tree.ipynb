# 04. Decision Tree Classifier

Non-linear decision tree classifier for romantic compatibility prediction using the full Cupid Feature Engineering pipeline.

```python
import sys
import os
import pandas as pd

# Ajout du dossier parent au PYTHONPATH
sys.path.append(os.path.abspath('..'))

# Importation des classes du projet
from src.data.preprocessing import CupidDataPreprocessor
from src.models.decision_tree import CupidDecisionTree
from src.evaluation.metrics import evaluate_classification_model, print_evaluation_summary

print("Modules du projet importés avec succès.")
```

```python
print("Préparation des données avec le pipeline complet de Feature Engineering...")
preprocessor = CupidDataPreprocessor(test_size=0.2, random_state=42)
raw_path = '../data/raw/cupid_algorithm_dataset.csv' if os.path.exists('../data/raw/cupid_algorithm_dataset.csv') else 'data/raw/cupid_algorithm_dataset.csv'
X_train, X_test, y_train, y_test = preprocessor.prepare_train_test(raw_path)

print(f"Nombre de features générées : {X_train.shape[1]}")
print(f"Forme de X_train : {X_train.shape}")
print(f"Forme de X_test  : {X_test.shape}")
```

```python
# Création et entraînement du modèle Decision Tree
model = CupidDecisionTree(max_depth=12, min_samples_split=10, random_state=42)
model.fit(X_train, y_train)
print("Modèle Arbre de Décision entraîné avec succès !")
```

```python
# Évaluation sur le jeu de test
metrics = evaluate_classification_model(model, X_test, y_test, 'Decision Tree')
print_evaluation_summary(metrics)
```
