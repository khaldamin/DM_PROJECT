# 04. Decision Tree Classifier

Non-linear decision tree classifier for matchmaking.

```python
import sys
import os
import pandas as pd

# On ajoute le dossier parent au chemin pour trouver le dossier 'src'
sys.path.append(os.path.abspath('..'))

# Importation des classes du projet
from src.models.decision_tree import CupidDecisionTree
from src.evluation.metrics import evaluate_classification_model, print_evaluation_summary
```

```python
# On charge directement les fichiers que vous venez de générer !
print("Chargement des données préparées...")
X_train = pd.read_csv('../data/processed/X_train.csv')
X_test = pd.read_csv('../data/processed/X_test.csv')
y_train = pd.read_csv('../data/processed/y_train.csv').values.ravel() # .values.ravel() évite les warnings
y_test = pd.read_csv('../data/processed/y_test.csv').values.ravel()

print(f"Forme de X_train : {X_train.shape}")
print(f"Forme de X_test : {X_test.shape}")
```

```python
# Création du modèle
model = CupidDecisionTree(max_depth=10, random_state=42)

# Entraînement
model.fit(X_train, y_train)
print("Modèle entraîné avec succès !")
```

```python
# Évaluation
metrics = evaluate_classification_model(model, X_test, y_test, 'Decision Tree')
print_evaluation_summary(metrics)
```
