Synthesizes comparative metrics across all four supervised learning algorithms and loads the best optimized model (Tuned XGBoost).

```python
import os
import json
import pandas as pd

results_path = '../results/model_comparison.csv' if os.path.exists('../results/model_comparison.csv') else 'results/model_comparison.csv'
comparison_df = pd.read_csv(results_path)
comparison_df
```

```python
summary_path = '../results/summary.json' if os.path.exists('../results/summary.json') else 'results/summary.json'
with open(summary_path, 'r', encoding='utf-8') as f:
    summary = json.load(f)

print("=== Meilleur Modèle Sélectionné ===")
print(f"Nom : {summary['best_model']}")
print(f"ROC-AUC : {summary['metrics']['ROC-AUC']:.4f} | Accuracy : {summary['metrics']['Accuracy']*100:.2f}% | F1-Score : {summary['metrics']['F1-Score']:.4f}")
print("Meilleurs Hyperparamètres :")
for k, v in summary['best_params'].items():
    print(f"  * {k}: {v}")
```
