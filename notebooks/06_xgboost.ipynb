# 06. XGBoost Gradient Boosting Classifier

State-of-the-art gradient boosted decision tree classifier providing sequential residual correction and high predictive performance.

```python
import sys
import os
import pandas as pd
sys.path.append(os.path.abspath('..'))

from src.data.preprocessing import CupidDataPreprocessor
from src.models.xgboost_model import CupidXGBoost
from src.evaluation.metrics import evaluate_classification_model, print_evaluation_summary

print("Modules XGBoost et évaluation importés avec succès.")
```

```python
print("Préparation des données avec CupidDataPreprocessor...")
preprocessor = CupidDataPreprocessor(test_size=0.2, random_state=42)
raw_path = '../data/raw/cupid_algorithm_dataset.csv' if os.path.exists('../data/raw/cupid_algorithm_dataset.csv') else 'data/raw/cupid_algorithm_dataset.csv'
X_train, X_test, y_train, y_test = preprocessor.prepare_train_test(raw_path)
print(f"X_train shape: {X_train.shape}, X_test shape: {X_test.shape}")
```

```python
print("Entraînement du modèle XGBoost (150 arbres)...")
model = CupidXGBoost(n_estimators=150, max_depth=6, learning_rate=0.08, subsample=0.8, colsample_bytree=0.8, n_jobs=-1, random_state=42)
model.fit(X_train, y_train)
print("Modèle XGBoost entraîné avec succès !")
```

```python
metrics = evaluate_classification_model(model, X_test, y_test, 'XGBoost')
print_evaluation_summary(metrics)
```
