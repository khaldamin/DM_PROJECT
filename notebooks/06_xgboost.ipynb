# 06. XGBoost Gradient Boosting Classifier

State-of-the-art gradient boosted decision tree classifier.

```python
import sys
sys.path.append('..')
from src.data.preprocessing import CupidDataPreprocessor
from src.models.xgboost_model import CupidXGBoost
from src.evluation.metrics import evaluate_classification_model, print_evaluation_summary
```

```python
preprocessor = CupidDataPreprocessor()
X_train, X_test, y_train, y_test = preprocessor.prepare_train_test('../data/raw/cupid_algorithm_dataset.csv')
```

```python
model = CupidXGBoost(n_estimators=150, max_depth=6, learning_rate=0.08, random_state=42)
model.fit(X_train, y_train)
metrics = evaluate_classification_model(model, X_test, y_test, 'XGBoost')
print_evaluation_summary(metrics)
```
