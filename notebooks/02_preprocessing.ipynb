```python
import os
import pandas as pd
import numpy as np

data_folder = "../data/raw"
csv_files = [f for f in os.listdir(data_folder) if f.lower().endswith(".csv")]
print("CSV files found:", csv_files)

# Use the first CSV (or replace with the exact file name if there are several)
df = pd.read_csv(os.path.join(data_folder, csv_files[0]))

print("Shape:", df.shape)
print("\nData types and non-null counts:")
df.info()
print("\nDuplicate rows:", df.duplicated().sum())
print("\nMissing values per column:")
print(df.isnull().sum().sort_values(ascending=False))
df.head()
```

```python
# --- Data quality checks ---
# 1. Missing values
missing = df.isnull().sum()
print("Columns with missing values:\n", missing[missing > 0])
print("Total missing:", missing.sum())

# 2. Duplicates
n_dup = df.duplicated().sum()
print("Duplicate rows:", n_dup)
df = df.drop_duplicates().reset_index(drop=True)

# 3. Duplicate pair IDs
print("Duplicate pair_id:", df["pair_id"].duplicated().sum())

# 4. Value ranges of numeric columns (look for impossible values)
print(df.describe().T[["min", "max", "mean"]])

# 5. Categorical columns and their unique values
cat_cols = df.select_dtypes(exclude="number").columns.tolist()
for c in cat_cols:
    print(f"\n{c}: {df[c].nunique()} unique ->", df[c].unique()[:10])
```

```python
TARGET = "compatible"   # la colonne à prédire

leakage_cols = ["compatibility_score", "relationship_longevity_months", "pair_id"]

print(df[TARGET].value_counts())
print(df[TARGET].value_counts(normalize=True).round(3))

X = df.drop(columns=leakage_cols + [TARGET])   # les indices
y = df[TARGET]                                  # la réponse à deviner
print("Features shape:", X.shape)
```

```python
# Colonnes texte à encoder
cat_cols = X.select_dtypes(exclude="number").columns.tolist()
print("Colonnes catégorielles :", cat_cols)

for c in cat_cols:
    print(f"{c}: {X[c].nunique()} valeurs ->", X[c].unique()[:8])

# One-hot encoding
X_encoded = pd.get_dummies(X, columns=cat_cols, drop_first=True, dtype=int)
print("\nAvant :", X.shape, "-> Après :", X_encoded.shape)
X_encoded.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Train :", X_train.shape, "| Test :", X_test.shape)
print("\nProportion dans train :\n", y_train.value_counts(normalize=True).round(3))
print("\nProportion dans test :\n", y_test.value_counts(normalize=True).round(3))
```

```python
from sklearn.preprocessing import StandardScaler

# Colonnes numériques d'origine (sans les colonnes 0/1 de l'encodage)
num_cols = X.select_dtypes(include="number").columns.tolist()
print("Colonnes à normaliser :", len(num_cols))

scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test_scaled[num_cols] = scaler.transform(X_test[num_cols])

X_train_scaled[num_cols].describe().T[["mean", "std"]].head()
```

```python
import os

os.makedirs("../data/processed", exist_ok=True)

X_train_scaled.to_csv("../data/processed/X_train.csv", index=False)
X_test_scaled.to_csv("../data/processed/X_test.csv", index=False)
y_train.to_csv("../data/processed/y_train.csv", index=False)
y_test.to_csv("../data/processed/y_test.csv", index=False)

print("Fichiers sauvegardés dans data/processed :", os.listdir("../data/processed"))
```
